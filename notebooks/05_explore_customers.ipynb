```python
customers["customer_unique_id"].duplicated().sum()
```

Some IDs labeled `customer_unique_id` repeat multiple times (3345 rows repeat a `customer_unique_id`) unlike `customer_id` which is unique per row.

```python
repeating_ids = customers["customer_unique_id"].value_counts()
repeating_ids = repeating_ids[repeating_ids > 1].index
sample_customer = repeating_ids[0]
customers[customers["customer_unique_id"] == sample_customer]
```

One `customer_unique_id` can have many rows: each order creates a new `customer_id`, while `customer_unique_id` stays the same person. Repeat buyers exist in the data.

```python
unique_customers = customers["customer_unique_id"].nunique()
unique_customers
```

96096 distinct customers placed 99441 orders, so some people ordered more than once.
