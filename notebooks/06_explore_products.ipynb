`olist_products_dataset.csv`

```python
import pandas as pd

products = pd.read_csv("../data/raw/olist_products_dataset.csv")
items = pd.read_csv("../data/raw/olist_order_items_dataset.csv") # used later
products.shape
```

32951 rows, 9 columns.

```python
products.head()
```

Columns cover the product ID, category name, number of characters in the product name and description, number of product photos, product weight (in grams), and dimensions (in centimeters).

```python
products.dtypes
```

Columns `product_id` and `product_category_name` are `str`, all other columns are `float64`.

```python
products.isnull().sum()
```

Most missing values are in product descriptive metadata (610 each). Weight and the three dimension columns have 2 nulls each. Those nulls are likely why the number columns are `float64` instead of `int64`.

```python
meta_nulls = products[products["product_category_name"].isnull()]
meta_nulls[["product_name_lenght", "product_description_lenght", "product_photos_qty"]].isnull().sum()
```

Products missing one descriptive metadata field are also missing all the other related fields.

```python
products["product_category_name"].nunique()
```

73 unique named product categories.

```python
products["product_category_name"].value_counts().head(10)
```

Top categories: bed/bath/table, sports/leisure, furniture/decoration. Portuguese names - translation table maps them to English.

```python
products["product_weight_g"].describe()
```

Product weights vary considerably, ranging from 0 to 40.4 kilograms. The average weight is 2.28 kilograms, while the median weight is 700 grams, suggesting that a small number of very heavy products increase the average.
